### Round A — Addition and multiplication


```python
_results.clear()
try:
    a, b = Value(2, label='a'), Value(3, label='b')
    c, d = a + b, a * b
    check("initial gradient is zero", a.grad == 0.0)
    check("addition data", c.data == 5)
    check("multiplication data", d.data == 6)
    check("addition parents", c._prev == {a,b})
    check("operation labels", c._op == '+' and d._op == '*')
except Exception as e: check("basic operations run", False, repr(e))
summary()
```
