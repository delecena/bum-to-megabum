## Goal

Connect symbolic derivatives to the finite-difference checks used in the source notebook.

Original expression:

$$f(a,b,c)=-a^3+\sin(3b)-\frac{1}{c}+b^{2.5}-\sqrt{a}$$

Before coding, write the three partial derivatives on paper.


Partial Derivates

$$\partial f/\partial a = -3a^2 - \frac{1}{2\sqrt{a}} \qquad |\qquad  \partial f/\partial b = 3\cos(3b) + 2.5b^{1.5}\qquad  |\qquad   \partial f/\partial c = \frac{1}{c^2}$$


```python
import math

_results = []

def check(name, condition, detail=""):
    ok = bool(condition)
    _results.append(ok)
    mark = "PASS" if ok else "FAIL"
    print(f"[{mark}] {name}" + (f" — {detail}" if detail else ""))

def close(a, b, tol=1e-6):
    return abs(a - b) <= tol

def summary():
    print(f"\nScore: {sum(_results)}/{len(_results)} tests passed")
```

```python
from math import sin, cos, sqrt

def f(a, b, c):
    return -a**3 + sin(3*b) - 1.0/c + b**2.5 - a**0.5

def analytic_grad(a, b, c):
    dfda = -3*(a**2) - (1/(2*sqrt(a)))
    dfdb = 3*cos(3*b) + 2.5*(b**1.5)
    dfdc = 1/c**2
    
    return dfda, dfdb, dfdc
```

Implement `analytic_grad` and predict the signs at `(2,3,4)` before testing.
